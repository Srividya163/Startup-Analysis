```python
import pandas as pd
import numpy as np
df=pd.read_csv('playstore_reviews.csv')
df
```

```python
df=df.drop(columns=['userImage','reviewId','repliedAt'])
df
```

```python
df['userName']=df['userName'].astype('string').str.strip()
df
```

```python
import re
import pandas as pd

# Make sure it's string type
df['content'] = df['content'].astype(str)

# Emoji regex covering most Unicode emojis
emoji_pattern = re.compile(
    "["
    "\U0001F600-\U0001F64F"  # emoticons
    "\U0001F300-\U0001F5FF"  # symbols & pictographs
    "\U0001F680-\U0001F6FF"  # transport & map symbols
    "\U0001F1E0-\U0001F1FF"  # flags
    "\U00002700-\U000027BF"  # dingbats
    "\U0001F900-\U0001F9FF"  # supplemental symbols
    "\U00002600-\U000026FF"  # misc symbols
    "\U00002B00-\U00002BFF"  # arrows etc
    "\U0001FA70-\U0001FAFF"  # new emojis
    "\U000025A0-\U000025FF"  # geometric shapes
    "]+", flags=re.UNICODE
)

# Remove emojis from every row
df['content'] = df['content'].apply(lambda x: emoji_pattern.sub('', x))

# Optional: remove extra spaces left behind
df['content'] = df['content'].str.replace(r'\s+', ' ', regex=True).str.strip()
df
```

```python
df=df.rename(columns=({'thumbsUpCount':'thumbs_up_count','reviewCreatedVersion':'review_version','at':'review_date','replyContent':'reply_content','appVersion':'app_version'}))
df
```

```python
df['review_version']=df['review_version'].astype('string').str.strip()
df['review_version']=df['review_version'].replace({"<NA>":"Unknown"}).fillna("Unkown")
print(df['review_version'].dtype)
df
```

```python
df['review_date']=pd.to_datetime(df['review_date'],errors='coerce')
df['review_date']=df['review_date'].dt.strftime('%d-%m-%Y')
df
```

```python
df['reply_content']=df['reply_content'].astype('string').str.capitalize().str.strip().fillna("No Reply")
df
```

```python
df['app_version']=df['app_version'].astype('string').str.strip().replace({"<NA>":"Unknown"}).fillna("Unknown")
df['company']=df['company'].astype('string').str.strip()
print(df.dtypes)
df.to_csv('playstore_reviews.csv',index=False)
```
